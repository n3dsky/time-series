#  Прогнозирование заказов такси

Компания «Чётенькое такси» собрала исторические данные о заказах такси в аэропортах. Чтобы привлекать больше водителей в период пиковой нагрузки, нужно спрогнозировать количество заказов такси на следующий час. Постройте модель для такого предсказания.

Значение метрики *RMSE* на тестовой выборке должно быть не больше 48.

Вам нужно:

1. Загрузить данные и выполнить их ресемплирование по одному часу.
2. Проанализировать данные.
3. Обучить разные модели с различными гиперпараметрами. Сделать тестовую выборку размером 10% от исходных данных.
4. Проверить данные на тестовой выборке и сделать выводы.


Данные лежат в файле `taxi.csv`. Количество заказов находится в столбце `num_orders` (от англ. *number of orders*, «число заказов»).

## Подготовка

### Импорт библиотек и настройки

Все импорты и общие константы собраны в одной ячейке в начале работы - так проще воспроизвести окружение и видно, какими инструментами мы пользуемся.

````python
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from lightgbm import LGBMRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit, cross_val_score
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import acf

warnings.filterwarnings('ignore')

# пути к данным: локальный и серверный (тренажёр)
LOCAL_PATH = 'datasets/taxi.csv'
SERVER_PATH = '/datasets/taxi.csv'

RANDOM_STATE = 42     # фиксируем случайность, чтобы результаты воспроизводились
TEST_SIZE = 0.1       # доля тестовой выборки по условию задачи
N_SPLITS = 5          # число блоков для кросс-валидации по времени
RMSE_TARGET = 48      # порог качества из условия задачи

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (14, 4)
pd.set_option('display.float_format', '{:.2f}'.format)
pd.set_option('display.max_colwidth', None)
````

### Загрузка данных

Читаем файл сразу «правильно» для временного ряда: колонку `datetime` делаем индексом (`index_col=0`) и разбираем её как дату (`parse_dates=[0]`). Тогда pandas сможет работать с ней как с временной шкалой - ресемплировать, срезать по датам, доставать час и день недели.

Конструкция `try / except` нужна, чтобы ноутбук запускался и локально, и в тренажёре, где файл лежит по пути `/datasets/taxi.csv`.

````python
try:
    data = pd.read_csv(LOCAL_PATH, index_col=0, parse_dates=[0])
except FileNotFoundError:
    data = pd.read_csv(SERVER_PATH, index_col=0, parse_dates=[0])

data.info()
print('\nПериод наблюдений:', data.index.min(), '-', data.index.max())
display(data.head())
````

### Проверка корректности временного ряда

Для временного ряда обычной проверки на пропуски и дубликаты мало. Дополнительно убедимся, что:

1. индекс монотонно возрастает (иначе «прошлое» и «будущее» перепутаны, и все лаговые признаки будут посчитаны неверно);
2. нет повторяющихся моментов времени;
3. шаг между наблюдениями везде одинаковый, то есть в ряду нет «дыр»;
4. значения неотрицательны - количество заказов не может быть меньше нуля.

````python
print('Тип индекса:                ', data.index.dtype)
print('Индекс возрастает:          ', data.index.is_monotonic_increasing)
print('Дубликатов дат:             ', data.index.duplicated().sum())
print('Пропусков в num_orders:     ', data['num_orders'].isna().sum())
print('Отрицательных значений:     ', (data['num_orders'] < 0).sum())

# шаг между соседними наблюдениями: если он везде одинаковый - пропущенных моментов нет
print('\nШаги между наблюдениями:')
print(data.index.to_series().diff().value_counts())
````

### Ресемплирование по одному часу

По условию задачи прогноз нужен на **следующий час**, поэтому переводим ряд из десятиминутного шага в часовой.

Агрегируем методом `sum()`, а не `mean()`: `num_orders` - это *количество* заказов, и заказы за час логично складывать. Среднее дало бы «среднее число заказов за 10 минут», то есть другую величину.

````python
# суммируем заказы внутри каждого часа
data = data.resample('1h').sum()

print('Размер после ресемплирования:', data.shape)
print('Разрывов в часовой сетке:', (data.index.to_series().diff().dropna() != pd.Timedelta('1h')).sum())
print('Часов в периоде (ожидаем):', (data.index.max() - data.index.min()).total_seconds() / 3600 + 1)

display(data.head())
display(data.describe())
````

**Вывод по разделу «Подготовка»**

- Исходные данные - 26 496 записей о числе заказов такси с 01.03.2018 по 31.08.2018 включительно, шаг измерения - 10 минут.
- Ряд корректен: индекс имеет тип `datetime64`, отсортирован по возрастанию, дубликатов дат и пропусков в `num_orders` нет, пропущенных десятиминуток тоже нет.
- После ресемплирования по одному часу получилось **4416 наблюдений** - ровно 184 суток × 24 часа, разрывов в часовой сетке нет.
- В среднем за час поступает ~84 заказа, медиана - 78. Разброс большой: от 0 до 462 заказов. Это говорит о выраженной неоднородности нагрузки, которую мы разберём в следующем разделе.

Данные готовы к анализу.

## Анализ

Цель раздела - понять, из чего состоит ряд: есть ли тренд, какие в нём повторяющиеся циклы и насколько он «шумный». От ответов зависит, какие признаки мы построим для моделей.

### Общая динамика ряда

Сырой почасовой ряд сильно колеблется, поэтому вместе с ним рисуем скользящие средние: за сутки (сглаживает суточную волну) и за неделю (показывает чистый тренд).

````python
fig, ax = plt.subplots(figsize=(14, 5))

data['num_orders'].plot(ax=ax, alpha=0.3, lw=0.8, label='исходный ряд (час)')
data['num_orders'].rolling(24).mean().plot(ax=ax, label='скользящее среднее, сутки')
data['num_orders'].rolling(24 * 7).mean().plot(ax=ax, lw=2.5, color='crimson', label='скользящее среднее, неделя')

ax.set_title('Число заказов такси, март - август 2018')
ax.set_ylabel('заказов в час')
ax.set_xlabel('')
ax.legend()
plt.show()

print('Среднее число заказов в час по месяцам:')
display(data['num_orders'].groupby(data.index.to_period('M')).mean().round(1))
````

**Что видно.** Ряд растёт весь период: недельное сглаживание (красная линия) поднимается почти монотонно, а средние по месяцам увеличиваются с 57.5 в марте до 127.7 в августе - рост в 2.2 раза. К августу подъём ускоряется. Значит, ряд нестационарен: его среднее зависит от времени. Для модели это означает, что тестовый период будет лежать выше всего, что она видела при обучении, и без опоры на недавнюю историю прогноз окажется систематически заниженным.

### Декомпозиция: тренд, сезонность, остаток

`seasonal_decompose` раскладывает ряд на три слагаемых: **тренд** (медленное изменение уровня), **сезонность** (повторяющийся цикл) и **остаток** (то, что не объяснили первые два). Параметр `period=24` означает, что мы ищем цикл длиной 24 часа - суточный.

Сезонная компонента по построению одинакова во всех сутках, поэтому выводить её на полгода бессмысленно: получится сплошная заливка. Покажем её на трёх сутках - так видна форма цикла.

````python
decomposed = seasonal_decompose(data['num_orders'], period=24)

fig, axes = plt.subplots(3, 1, figsize=(14, 9))

decomposed.trend.plot(ax=axes[0])
axes[0].set_title('Тренд')

# сезонная компонента повторяется каждые сутки, поэтому показываем её на трёх днях
decomposed.seasonal['2018-03-01':'2018-03-03'].plot(ax=axes[1], marker='.')
axes[1].set_title('Сезонность (суточный цикл, показаны трое суток)')

decomposed.resid.plot(ax=axes[2], alpha=0.6)
axes[2].set_title('Остаток')

for ax in axes:
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
````

**Что видно.** Разложение подтверждает две вещи. Тренд (верхняя панель) устойчиво растёт, к августу заметно ускоряясь. Сезонная компонента (средняя панель) имеет чёткую суточную форму с провалом под утро и пиком в ночные часы. Остаток (нижняя панель) колеблется вокруг нуля без видимой структуры, но его размах увеличивается к концу периода: чем выше поток заказов, тем больше абсолютные отклонения. Отсюда практический вывод - на тестовой части (август) RMSE будет заведомо выше, чем на весенних данных, просто из-за масштаба ряда.

### Профили сезонности

Декомпозиция показала, что суточный цикл есть. Теперь измерим его количественно: усредним заказы по каждому часу суток и по каждому дню недели за весь период. Сравнение размаха этих двух профилей подскажет, какая сезонность важнее для модели.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

by_hour = data.groupby(data.index.hour)['num_orders'].mean()
by_hour.plot(ax=axes[0], marker='o')
axes[0].set_title('Средние заказы по часам суток')
axes[0].set_xlabel('час суток')
axes[0].set_ylabel('заказов в час')
axes[0].set_xticks(range(0, 24, 2))

by_weekday = data.groupby(data.index.dayofweek)['num_orders'].mean()
by_weekday.index = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']
by_weekday.plot(kind='bar', ax=axes[1], rot=0)
axes[1].set_title('Средние заказы по дням недели')
axes[1].set_xlabel('')
axes[1].set_ylabel('заказов в час')

plt.tight_layout()
plt.show()

print('Разброс средних по часам суток:  от', round(by_hour.min()), 'до', round(by_hour.max()))
print('Разброс средних по дням недели:  от', round(by_weekday.min()), 'до', round(by_weekday.max()))
````

**Что видно.** Суточный профиль выражен резко: минимум приходится на 06:00 (около 25 заказов), максимум - на полночь (около 144), заметен вечерний подъём к 16:00. Размах средних по часам - от 25 до 144.

Профиль по дням недели на первый взгляд выглядит плоским: от 77 заказов во вторник до 91 в понедельник и пятницу. Но усреднение по всем суткам сглаживает картину: складывая вместе ночные пики и утренние провалы, мы теряем структуру. Проверим это на следующем графике.

### Совместная сезонность: час суток и день недели

Средние по дням недели показали разброс всего 77-91 заказ, и это легко принять за слабую недельную сезонность. Однако час суток и день недели действуют **совместно**: понедельничная полночь и субботняя полночь могут отличаться сильнее, чем понедельник отличается от субботы «в среднем за сутки».

Построим тепловую карту: по вертикали час суток, по горизонтали день недели, в ячейке - среднее число заказов.

````python
hourly_weekly = data.copy()
hourly_weekly['hour'] = hourly_weekly.index.hour
hourly_weekly['dayofweek'] = hourly_weekly.index.dayofweek

pivot = hourly_weekly.pivot_table(index='hour', columns='dayofweek', values='num_orders', aggfunc='mean')
pivot.columns = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(pivot, cmap='YlOrRd', ax=ax, cbar_kws={'label': 'среднее число заказов'})
ax.set_title('Средние заказы: час суток x день недели')
ax.set_ylabel('час суток')
ax.set_xlabel('')
plt.show()

print(f'Размах средних по часам суток:        {pivot.mean(axis=1).min():.0f} - {pivot.mean(axis=1).max():.0f}')
print(f'Размах средних по дням недели:        {pivot.mean(axis=0).min():.0f} - {pivot.mean(axis=0).max():.0f}')
print(f'Размах по всем ячейкам «час x день»:  {pivot.min().min():.0f} - {pivot.max().max():.0f}')
````

**Что видно.** Разброс по отдельным ячейкам «час x день недели» существенно шире, чем разброс по каждому измерению в отдельности. Ночные пики заметно различаются по дням недели, а утренние провалы почти одинаковы - то есть день недели влияет неравномерно по часам.

Вывод: недельную сезонность нельзя списывать со счетов. Час и день недели работают как пара признаков, и модель должна получить оба, а вместе с ними - лаг, отвечающий за недельный цикл.

### Автокорреляция: какие лаги действительно значимы

Вместо того чтобы подбирать глубину истории перебором, спросим сами данные. Автокорреляционная функция (ACF) показывает, насколько ряд похож сам на себя при сдвиге на *k* часов назад. Высокое значение на лаге *k* означает, что значение *k* часов назад - хороший предиктор текущего.

Синяя заливка на графике - порог статистической значимости: если столбик внутри неё, корреляция уже не заслуживает доверия.

````python
fig = plot_acf(data['num_orders'], lags=190)
fig.set_size_inches(14, 5)
plt.ylim(-0.1, 1.1)
plt.title('Автокорреляция числа заказов')
plt.xlabel('лаг, часов')
plt.show()

acf_values = pd.Series(acf(data['num_orders'], nlags=190)).drop(0)
acf_values.index.name = 'лаг, часов'
print('Десять самых сильных лагов:')
display(acf_values.sort_values(ascending=False).head(10).rename('автокорреляция').to_frame())
````

**Что видно.** Автокорреляция не затухает плавно, а даёт отчётливые пики на лагах, кратным суткам: 24, 48, 72, 96, 120, 144 и 168. Все они далеко за порогом значимости.

Самый сильный лаг во всём ряду - **168 часов, то есть ровно неделя** (0.72). Он обходит и суточный лаг 24 (0.68), и соседний час (0.62). Это прямое численное подтверждение того, что показала тепловая карта: заказы в конкретный час больше всего похожи на заказы в тот же час той же дневной недели, а не на заказы часом ранее.

Отсюда набор лагов для модели получается без всякого перебора: **1-24** (недавняя история и полный суточный цикл) плюс **48, 72, 96, 120, 144, 168** (суточные пики вплоть до недельного).

## Обучение

### От прогноза ряда к задаче регрессии

Обычные модели машинного обучения не умеют работать с рядом напрямую: им нужна таблица «признаки -> ответ». Превратим ряд в такую таблицу.

**Ответ** (`num_orders` в строке за час *t*) - то, что предсказываем. **Признаки** - всё, что известно к моменту *t*, то есть информация о часах *до* него:

| Признак | Что это | Обоснование из раздела «Анализ» |
|---|---|---|
| `hour` | час суток (0-23) | размах средних по часам 25-144, сильнейший внутрисуточный сигнал |
| `dayofweek` | день недели (0-6) | вместе с часом задаёт положение в недельном цикле |
| `lag_1 … lag_24` | заказы 1-24 часа назад | недавняя история и полный суточный цикл |
| `lag_48, lag_72, lag_96, lag_120, lag_144, lag_168` | заказы 2-7 суток назад в тот же час | пики автокорреляции, кратные суткам; лаг 168 - самый сильный из всех |
| `rolling_mean` | среднее за последние 12 часов | сглаженный локальный уровень, помогает улавливать тренд |

Ключевое правило - **не заглядывать в будущее**. Каждый лаг получается методом `shift(lag)`, то есть сдвигом ряда назад, а скользящее среднее считается с `shift()` *перед* `rolling()`. Без этого сдвига в среднее попало бы текущее значение - то самое, которое мы предсказываем, - и модель показала бы фантастическое качество на обучении и провалилась на тесте. Это классическая утечка целевой переменной.

Признак «месяц» намеренно не добавляем: в тесте встречается только август, и модель, привыкшая к номерам месяцев 3-7, не смогла бы разумно экстраполировать значение 8.

После сдвигов у первых строк лаги пустые - их удаляем через `dropna()`.

````python
def make_features(df, lags, rolling_mean_size):
    """Строит признаки для прогноза числа заказов на один час вперёд.

    lags - список лагов, выбранный по графику автокорреляции.
    Все признаки считаются только по прошлым значениям ряда,
    поэтому «заглядывания в будущее» не происходит.
    """
    features = df.copy()

    # календарные признаки: положение часа в сутках и в неделе
    features['hour'] = features.index.hour
    features['dayofweek'] = features.index.dayofweek

    # значения ряда, сдвинутые назад на указанные лаги
    for lag in lags:
        features[f'lag_{lag}'] = features['num_orders'].shift(lag)

    # локальный уровень ряда; shift() обязателен, иначе в среднее
    # попадёт текущее значение - то самое, которое мы предсказываем
    features['rolling_mean'] = features['num_orders'].shift().rolling(rolling_mean_size).mean()

    # первые строки остались без лагов - убираем их
    return features.dropna()


# пример: как выглядит результат на небольшом наборе лагов
display(make_features(data, lags=[1, 2, 24], rolling_mean_size=12).head())
````

### Разделение выборок и схема валидации

Два правила, специфичных именно для временных рядов:

**1. Разделяем по времени, а не случайно.** Тестом становятся *последние* 10% часов (441 час - вторая половина августа). Перемешивать нельзя: если в обучение попадут «будущие» часы, модель фактически подсмотрит ответ, метрика окажется красивой, а в реальной работе модель провалится.

**2. Кросс-валидация - тоже по времени.** Обычный `KFold` тасует данные и нарушает правило выше. Используем `TimeSeriesSplit`: он строит несколько последовательных блоков, где обучение всегда идёт по прошлому, а проверка - по следующему за ним отрезку:

```
фолд 1:  [обучение]  [валидация]
фолд 2:  [  обучение   ]  [валидация]
фолд 3:  [     обучение    ]  [валидация]
```

Это честная имитация того, как модель будет работать в проде: учимся на накопленной истории, предсказываем ближайшее будущее. Тестовую выборку при подборе гиперпараметров не трогаем вообще - она остаётся нетронутой до финальной проверки.

````python
TEST_LEN = int(len(data) * TEST_SIZE)   # 10% от исходного часового ряда
tscv = TimeSeriesSplit(n_splits=N_SPLITS)


def split_data(features, test_len):
    """Делит выборку по времени: тест - последние test_len часов, без перемешивания."""
    train, test = features.iloc[:-test_len], features.iloc[-test_len:]
    return (
        train.drop('num_orders', axis=1), train['num_orders'],
        test.drop('num_orders', axis=1), test['num_orders'],
    )


def cv_rmse(model, X, y, cv=tscv):
    """Средний RMSE модели по блокам кросс-валидации для временного ряда."""
    scores = cross_val_score(model, X, y, cv=cv, scoring='neg_root_mean_squared_error')
    return -scores.mean()


print('Размер тестовой выборки:', TEST_LEN, 'часов')
````

### Набор лагов по автокорреляции

Автокорреляционный анализ из предыдущего раздела уже ответил на вопрос, какие лаги значимы, поэтому подбирать их перебором не нужно. Берём:

- **лаги 1-24** - недавняя история и полный суточный цикл. Лаг 1 (0.62) отражает плавность ряда, лаг 24 (0.68) - суточную сезонность;
- **лаги 48, 72, 96, 120, 144, 168** - пики автокорреляции, кратные суткам. Все они значимы, а недельный лаг 168 (0.72) - сильнейший во всём ряду;
- **скользящее среднее с окном 12 часов** - сглаженный уровень за половину суток.

Промежуточные лаги (например, 100 или 130) в набор не включаем: на графике ACF они лежат заметно ниже суточных пиков и лишь добавили бы модели шума.

````python
# лаги выбраны по графику автокорреляции из раздела «Анализ»
DAILY_LAGS = [48, 72, 96, 120, 144, 168]     # пики, кратные суткам, вплоть до недельного
LAGS = list(range(1, 25)) + DAILY_LAGS
ROLLING_MEAN_SIZE = 12

print('Лаги:', LAGS)
print('Всего лагов:', len(LAGS), '  окно скользящего среднего:', ROLLING_MEAN_SIZE)
````

Собираем итоговые выборки, на которых будем обучать модели.

````python
features = make_features(data, LAGS, ROLLING_MEAN_SIZE)
X_train, y_train, X_test, y_test = split_data(features, TEST_LEN)

print('Число признаков:', X_train.shape[1])
print(f'Обучающая выборка: {len(X_train):>5} ч   {X_train.index.min()} - {X_train.index.max()}')
print(f'Тестовая выборка:  {len(X_test):>5} ч   {X_test.index.min()} - {X_test.index.max()}')
print(f'Доля теста от исходного ряда: {len(X_test) / len(data):.1%}')

print('\nБлоки кросс-валидации внутри обучающей выборки:')
for i, (train_index, val_index) in enumerate(tscv.split(X_train), 1):
    print(f'  фолд {i}: обучение {len(train_index):>5} ч  ->  валидация {len(val_index):>4} ч '
          f'({X_train.index[val_index[0]]:%d.%m.%Y} - {X_train.index[val_index[-1]]:%d.%m.%Y})')

display(X_train.head())
````

### Бейзлайн: решение без машинного обучения

Прежде чем обучать модели, зафиксируем качество самого простого правила, не требующего никакого ML. Это проверка на адекватность: если сложная модель не обгонит одну строчку кода, внедрять её бессмысленно - бизнес не должен платить за инфраструктуру, которая не даёт выигрыша.

Для временных рядов такой бейзлайн - прогноз предыдущим значением. Но каким именно?

- **«как в прошлый час»** (`lag_1`) - привычный вариант, однако автокорреляция на лаге 1 равна 0.62;
- **«как в тот же час неделю назад»** (`lag_168`) - автокорреляция 0.72, самая высокая в ряду.

Раз ряд сильнее всего похож на себя недельной давности, честный бейзлайн - именно недельный. Считаем оба, но ориентироваться будем на более сильный: побить слабого соперника легко, а планка должна быть настоящей.

````python
def naive_cv_rmse(X, y, cv, lag_column):
    """RMSE прогноза, который просто повторяет значение ряда из указанного лага."""
    scores = []
    for _, val_index in cv.split(X):
        prediction = X.iloc[val_index][lag_column]
        scores.append(mean_squared_error(y.iloc[val_index], prediction) ** 0.5)
    return np.mean(scores)


baselines = {
    'Бейзлайн: как неделю назад (lag_168)': 'lag_168',
    'Бейзлайн: как в прошлый час (lag_1)': 'lag_1',
}

# сюда будем складывать результаты всех решений
cv_results = []
for name, column in baselines.items():
    rmse = naive_cv_rmse(X_train, y_train, tscv, column)
    cv_results.append({
        'Модель': name,
        'RMSE (кросс-валидация)': rmse,
        'Лучшие гиперпараметры': '-',
    })
    print(f'{name}: RMSE = {rmse:.2f}')
````

### Обучение моделей и подбор гиперпараметров

Сравним четыре модели двух разных природ:

| Модель | Идея | Что подбираем |
|---|---|---|
| **Линейная регрессия** | взвешенная сумма признаков | нечего подбирать |
| **Гребневая регрессия (Ridge)** | то же, но с штрафом за большие веса | силу регуляризации `alpha` |
| **Случайный лес** | усреднение множества независимых деревьев | число деревьев, глубина, минимальный размер листа |
| **LightGBM** | градиентный бустинг: деревья учатся на ошибках предыдущих | число деревьев, скорость обучения, глубина, число листьев |

Линейных моделей здесь две не случайно. У нашего ряда две особенности, которые играют против деревьев:

1. **Зависимость близка к аддитивной.** Уровень заказов складывается из тренда, суточной волны и недельной поправки. Линейная модель выражает такую сумму одной формулой, а дереву приходится дробить пространство признаков на множество прямоугольных областей.
2. **Ряд растёт, и тест лежит выше всего, что было в обучении.** Дерево физически не может предсказать значение выше максимума, встреченного при обучении: его прогноз - это среднее по листу. Линейная модель экстраполирует свободно.

Модели и их сетки гиперпараметров храним в одном словаре и перебираем циклом - так код не дублируется, а добавление новой модели стоит трёх строк.

`GridSearchCV` перебирает все комбинации параметров, для каждой считает RMSE по нашим пяти временным блокам и возвращает лучшую. Метрика `neg_root_mean_squared_error` - это RMSE со знаком минус (sklearn всегда максимизирует метрику), обратно переводим сменой знака.

````python
models = {
    'Линейная регрессия': (
        LinearRegression(),
        {},                                   # у модели нет гиперпараметров для подбора
    ),
    'Гребневая регрессия (Ridge)': (
        Ridge(random_state=RANDOM_STATE),
        {'alpha': [0.1, 1, 10, 100]},
    ),
    'Случайный лес': (
        RandomForestRegressor(random_state=RANDOM_STATE),
        {
            'n_estimators': [100, 300],
            'max_depth': [5, 10, None],
            'min_samples_leaf': [1, 5],
        },
    ),
    'LightGBM': (
        LGBMRegressor(random_state=RANDOM_STATE, verbose=-1),
        {
            'n_estimators': [100, 300, 500],
            'learning_rate': [0.03, 0.1],
            'num_leaves': [15, 31],
            'max_depth': [3, 6, -1],
        },
    ),
}

fitted_models = {}

for name, (estimator, param_grid) in models.items():
    search = GridSearchCV(
        estimator,
        param_grid,
        scoring='neg_root_mean_squared_error',
        cv=tscv,
        n_jobs=-1,
    )
    search.fit(X_train, y_train)

    fitted_models[name] = search.best_estimator_
    cv_results.append({
        'Модель': name,
        'RMSE (кросс-валидация)': -search.best_score_,
        'Лучшие гиперпараметры': str(search.best_params_) if search.best_params_ else '-',
    })
    print(f'{name}: RMSE = {-search.best_score_:.2f}, параметры: {search.best_params_}')
````

````python
cv_results = pd.DataFrame(cv_results).sort_values('RMSE (кросс-валидация)').reset_index(drop=True)
display(cv_results)

# лучшей считаем модель, а не бейзлайн, поэтому бейзлайны исключаем из выбора
best_name = cv_results.loc[~cv_results['Модель'].str.startswith('Бейзлайн'), 'Модель'].iloc[0]
best_model = fitted_models[best_name]
print('Лучшая модель по кросс-валидации:', best_name)
````

Никакая настройка гиперпараметров такого эффекта не даёт: разброс между всеми четырьмя моделями составляет всего 0.3, тогда как правильный набор лагов улучшил результат на несколько единиц.
2. **Оба бейзлайна побиты с запасом.** Прогноз «как неделю назад» (26.14) заметно сильнее прогноза «как в прошлый час» (37.29) - ещё одно подтверждение того, что недельный цикл в этих данных главный. Именно недельный бейзлайн и служит настоящей планкой.
3. **Линейные модели впереди деревьев.** Линейная и гребневая регрессии дали 22.03, LightGBM - 22.30, случайный лес - 22.32. На кросс-валидации разрыв невелик, но на тестовой выборке он раскроется гораздо сильнее (см. следующий раздел).
4. **Регуляризация ничего не изменила:** Ridge с `alpha=100` показал ровно тот же результат, что и обычная линейная регрессия. Значит, при 33 признаках и 3807 наблюдениях линейная модель и без штрафа не переобучается. Дальше используем Ridge как формально лучшую по кросс-валидации, но выводы для обеих моделей одинаковы.
5. **LightGBM выбрал очень скромные деревья** (`max_depth=3`, `num_leaves=15`, `learning_rate=0.03`). Бустинг сам ограничил сложность: при аддитивной структуре данных глубокие деревья только вредят.

## Тестирование

До этого момента тестовая выборка ни разу не участвовала ни в обучении, ни в подборе параметров - мы работали только с кросс-валидацией внутри обучающей части. Теперь проверим решения на этих отложенных 441 часах.

Прогоняем по тесту **все** обученные модели и **оба бейзлайна**. Два ориентира:

1. формальное требование задачи - RMSE не больше 48;
2. содержательный ориентир - бейзлайн по 168-му лагу. Модель имеет смысл внедрять, только если она обгоняет решение без машинного обучения.

Выбор модели при этом уже сделан по кросс-валидации, тест используется только для оценки.

````python
test_results = []

for name, column in baselines.items():
    test_results.append({
        'Модель': name,
        'RMSE (тест)': mean_squared_error(y_test, X_test[column]) ** 0.5,
    })

for name, model in fitted_models.items():
    test_results.append({
        'Модель': name,
        'RMSE (тест)': mean_squared_error(y_test, model.predict(X_test)) ** 0.5,
    })

final_results = (
    cv_results
    .merge(pd.DataFrame(test_results), on='Модель')
    .sort_values('RMSE (тест)')
    .reset_index(drop=True)
)
display(final_results[['Модель', 'RMSE (кросс-валидация)', 'RMSE (тест)', 'Лучшие гиперпараметры']])

best_rmse = final_results.loc[final_results['Модель'] == best_name, 'RMSE (тест)'].iloc[0]
baseline_name = 'Бейзлайн: как неделю назад (lag_168)'
baseline_rmse = final_results.loc[final_results['Модель'] == baseline_name, 'RMSE (тест)'].iloc[0]

print(f'\nВыбранная по кросс-валидации модель: {best_name}')
print(f'RMSE на тесте:                  {best_rmse:.2f}')
print(f'Бейзлайн (неделю назад):        {baseline_rmse:.2f}')
print(f'Порог из условия задачи:        {RMSE_TARGET}')
print(f'\nВыигрыш у бейзлайна: {1 - best_rmse / baseline_rmse:.1%}')
print('Порог задачи пройден' if best_rmse <= RMSE_TARGET else 'Порог задачи НЕ пройден')
print('Модель обгоняет решение без ML, её можно рекомендовать к внедрению'
      if best_rmse < baseline_rmse else
      'Модель не обгоняет решение без ML, внедрять её нецелесообразно')
````

### Прогноз против факта

Метрика говорит «насколько», а график - «где именно» модель ошибается. Смотрим на весь тестовый период и на первую неделю крупным планом.

````python
predictions = pd.Series(best_model.predict(X_test), index=X_test.index)

fig, axes = plt.subplots(2, 1, figsize=(14, 9))

y_test.plot(ax=axes[0], lw=1, label='факт')
predictions.plot(ax=axes[0], lw=1, label='прогноз')
axes[0].set_title(f'Вся тестовая выборка: факт и прогноз ({best_name})')
axes[0].set_ylabel('заказов в час')
axes[0].legend()

y_test[:24 * 7].plot(ax=axes[1], marker='.', label='факт')
predictions[:24 * 7].plot(ax=axes[1], marker='.', label='прогноз')
axes[1].set_title('Первая неделя тестовой выборки крупным планом')
axes[1].set_ylabel('заказов в час')
axes[1].legend()

for ax in axes:
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
````

### Анализ ошибок

RMSE - одно число, оно не показывает, *как именно* модель ошибается. Посмотрим на остатки (факт - прогноз): нет ли систематического смещения, растёт ли ошибка со временем, симметрично ли распределены промахи.

````python
residuals = y_test - predictions

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

residuals.plot(ax=axes[0], lw=0.9)
axes[0].axhline(0, color='crimson', lw=1)
axes[0].set_title('Ошибки прогноза во времени')
axes[0].set_ylabel('факт - прогноз')
axes[0].set_xlabel('')

residuals.plot(kind='hist', bins=40, ax=axes[1])
axes[1].axvline(0, color='crimson', lw=1)
axes[1].set_title('Распределение ошибок')
axes[1].set_xlabel('факт - прогноз')

plt.tight_layout()
plt.show()

print(f'Средняя ошибка (смещение):     {residuals.mean():+.2f} заказа')
print(f'Средняя абсолютная ошибка MAE: {residuals.abs().mean():.2f} заказа')
print(f'Средний уровень ряда на тесте: {y_test.mean():.1f} заказа в час')
print(f'Относительная ошибка:          {residuals.abs().mean() / y_test.mean():.1%}')
````

### На что модель опирается

Посмотрим, какие признаки внесли наибольший вклад. Для линейных моделей это коэффициенты (со знаком: положительный означает, что рост признака поднимает прогноз), для деревьев - встроенная важность признаков.

Это проверка на здравый смысл: если в топе окажутся признаки, отражающие суточный и недельный циклы, значит модель выучила ровно то, что мы увидели в разделе «Анализ».

````python
if hasattr(best_model, 'feature_importances_'):
    weights = pd.Series(best_model.feature_importances_, index=X_test.columns)
    label = 'важность признака'
else:
    weights = pd.Series(best_model.coef_, index=X_test.columns)
    label = 'коэффициент модели'

top_features = weights.reindex(weights.abs().sort_values(ascending=False).index).head(15)

fig, ax = plt.subplots(figsize=(10, 5))
top_features.sort_values().plot(kind='barh', ax=ax)
ax.axvline(0, color='black', lw=0.8)
ax.set_title(f'15 самых влиятельных признаков ({best_name})')
ax.set_xlabel(label)
plt.show()

display(top_features.rename(label).to_frame())
````
